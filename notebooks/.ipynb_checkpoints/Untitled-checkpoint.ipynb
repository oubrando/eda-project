```python
#load arrivals datset
arrivals_path = Path("../data/Raw/Arrivals_AA_DFW_2016_2025.csv")
arrivals = pd.read_csv(arrivals_path, skiprows=7,parse_dates=["Date (MM/DD/YYYY)"])

arrivals.head()
```

```python
#load departures datset
departures_path = Path("../data/Raw/Departures_AA_DFW_2016_2025.csv")
departures = pd.read_csv(departures_path, skiprows=7,parse_dates=["Date (MM/DD/YYYY)"])

departures.head()
```

```python
#Ensure we only have "AA" as the carrier
print("Arrivals Carrier Codes:",arrivals["Carrier Code"].value_counts())
print("Departures Cardrier Codes:", departures["Carrier Code"].value_counts())
```

```python
#Drop Carrier Code and Tail Number Columns
departures.drop(columns=["Carrier Code", "Tail Number"],inplace=True)
arrivals.drop(columns=["Carrier Code", "Tail Number"],inplace=True)
```
