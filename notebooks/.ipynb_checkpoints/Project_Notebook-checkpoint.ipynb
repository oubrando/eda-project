### Validation

```python
#Validate time-of-day fields by converting to datetime objects, coercing errors to generate NaN's
def invalid_times(tcol):
    converted = pd.to_datetime(tcol, format="%H:%M", errors="coerce")
    return converted.isna().sum()

print("Missing/Improperly Formatted Departure Times of Day")
departure_tcols = ["Scheduled departure time", "Actual departure time", "Wheels-off time"]
for tcol in departure_tcols:
    print("\t",tcol,":",invalid_times(departures[tcol]))

print("Missing/Improperly Formatted Arrival Times of Day")
arrival_tcols = ["Scheduled Arrival Time", "Actual Arrival Time", "Wheels-on Time"]
for tcol in arrival_tcols:
    print("\t",tcol,":",invalid_times(arrivals[tcol]))

```

```python
# Validate date fields with a similar approach
arrivals_date_col = pd.to_datetime(arrivals["Date (MM/DD/YYYY)"], errors="coerce")
print("Missing Arrivals Dates:",arrivals_date_col.isna().sum())

departures_date_col = pd.to_datetime(departures["Date (MM/DD/YYYY)"], errors="coerce")
print("Missing Departures Dates:",departures_date_col.isna().sum())
```

```python
#Validate duration fields are reasonable (between 2 hours early and no more than 24 hours late)
def invalid_duration(col, min_val=-120, max_val=1440):
    mask = col.between(min_val, max_val, inclusive="both")
    return (~mask).sum()

arrivals_duration_cols = ["Scheduled Elapsed Time (Minutes)", 
    "Actual Elapsed Time (Minutes)",
    "Arrival Delay (Minutes)",
    "Taxi-In time (Minutes)",
    "Delay Carrier (Minutes)",
    "Delay Weather (Minutes)",
    "Delay National Aviation System (Minutes)",
    "Delay Security (Minutes)",
    "Delay Late Aircraft Arrival (Minutes)"
    ]

print("="*20,"Arrival Durations Validation","="*20)
for duration_col in arrivals_duration_cols:
    print("Extreme durations in",duration_col,invalid_duration(arrivals[duration_col]))
    
print("="*20,"Departure Durations Validation","="*20)

departures_duration_cols = ["Scheduled elapsed time (Minutes)", 
    "Actual elapsed time (Minutes)",
    "Departure delay (Minutes)",
    "Taxi-Out time (Minutes)",
    "Delay Carrier (Minutes)",
    "Delay Weather (Minutes)",
    "Delay National Aviation System (Minutes)",
    "Delay Security (Minutes)",
    "Delay Late Aircraft Arrival (Minutes)"
    ]
for duration_col in departures_duration_cols:
    print("Extreme durations in",duration_col,invalid_duration(departures[duration_col]))
```

### Duplicate Checks

```python
# Check for duplicates on key columns

key_departure_cols = ["Date (MM/DD/YYYY)","Flight Number","Tail Number","Scheduled departure time"]
dep_duplicate_mask = departures.duplicated(subset=key_departure_cols, keep=False)
print("Potential duplicates in departures:", dep_duplicate_mask.sum())

key_arrival_cols = ["Date (MM/DD/YYYY)","Flight Number","Tail Number","Scheduled Arrival Time"]
arr_duplicate_mask = arrivals.duplicated(subset=key_arrival_cols, keep=False)
print("Potential duplicates in arrivals:", arr_duplicate_mask.sum())
```

### Outlier Checks

```python
#Check for outliers in float64 columns

def iqr_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    mask = (series < lower) | (series > upper)
    return mask
    
print("="*20,"Arrivals Outliers","="*20)
arr_float_cols = arrivals.select_dtypes(include="float64").columns
for col in arr_float_cols:
    outliers_mask = iqr_outliers(arrivals[col])
    outliers_pct = outliers_mask.sum() / len(arrivals[col]) * 100
    print("Outliers Percentage in",col,":",round(outliers_pct,2),"%")

print("="*20,"Departures Outliers","="*20)
dep_float_cols = departures.select_dtypes(include="float64").columns
for col in dep_float_cols:
    outliers_mask = iqr_outliers(departures[col])
    outliers_pct = outliers_mask.sum() / len(departures[col]) * 100
    print("Outliers Percentage in",col,":",round(outliers_pct,2),"%")
```

### Data Quality Assessment
Based on checks of missing data, duplicate row, reasonableness according to domain expectations, and outlier detections, these datasets can be trusted with no cleaning.

## Phase 3: Cleaning Decisions
Based on the above data quality assessment, we do not have to perform any significant cleaning that will jeopardize the truth.

To make later analyses more efficient, we will concatenate the two datasets, convert time-of-day fields to datetime fields, and drop unneeded columns.

```python
# To merge arrivals and departures, change column names to a common scheme
departures_mod = departures.assign(
    flight_event="Departure", 
    scheduled_time=departures["Scheduled departure time"],
    actual_time=departures["Actual departure time"],
    wheels_time=departures["Wheels-off time"],      
    taxi_time=departures["Taxi-Out time (Minutes)"],
    delay=departures['Departure delay (Minutes)']
)

arrivals_mod = arrivals.assign(
    flight_event="Arrivals", 
    scheduled_time=arrivals["Scheduled Arrival Time"],
    actual_time=arrivals["Actual Arrival Time"],
    wheels_time=arrivals["Wheels-on Time"],      
    taxi_time=arrivals["Taxi-In time (Minutes)"],
    delay=arrivals['Arrival Delay (Minutes)']
)

```

```python
# Drop the columns that have been renamed
dep_drop_cols = [
    "Scheduled departure time",
    "Actual departure time",
    "Taxi-Out time (Minutes)",
    "Wheels-off time",
    'Departure delay (Minutes)'
]

arr_drop_cols = [
    "Scheduled Arrival Time",
    "Actual Arrival Time",
    "Taxi-In time (Minutes)",
    "Wheels-On time",
    'Arrival Delay (Minutes)'
]

departures_mod = departures_mod.drop(columns=[c for c in dep_drop_cols if c in departures_mod.columns])
arrivals_mod= departures_mod.drop(columns=[c for c in arr_drop_cols if c in departures_mod.columns])
```

```python
#Concatenate the two data frames
combined = pd.concat([departures_mod, arrivals_mod], ignore_index=True, sort=False)
combined.head()
```

```python
combined.tail()
```

```python
combined.info()
```

```python
#Ensure we only have "AA" as the carrier
print("Arrivals Carrier Codes:",combined["Carrier Code"].value_counts())
```

```python
#Drop Carrier Code and Tail Number Columns since those are not needed for this analysis
combined.drop(columns=["Carrier Code", "Tail Number"],inplace=True)
```

```python
#Check for any lingering missing values
combined.isna().sum()
```

```python
# Combine the time columns with the date to make valide datetime objects

#Change the Date column to datetime objects
combined["Date (MM/DD/YYYY)"] = pd.to_datetime(combined["Date (MM/DD/YYYY)"])

combined_mod = combined.copy() #Copying to prevent overwriting the original

tcols = ["scheduled_time", "actual_time", "wheels_time"]
for tcol in tcols:
    combined_mod[tcol + "_dt"] = pd.to_datetime(
        combined_mod["Date (MM/DD/YYYY)"].dt.strftime("%Y-%m-%d") + " " + combined_mod[tcol].astype(str),
        format="%Y-%m-%d %H:%M",
        errors="coerce" #Deal with data not in ##:## by replacing with NaN
    )
# Since the above might have resulted in NaNs, drop them again
combined_mod.dropna(inplace=True)

# # Drop the appropriate columns now that they are not needed
combined_mod.drop(columns=tcols, inplace=True)
combined_mod.drop(columns=["Date (MM/DD/YYYY)"], inplace=True)
```
